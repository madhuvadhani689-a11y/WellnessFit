```python
from flask import Flask, request
from threading import Thread

app = Flask(__name__)

@app.route("/")
def home():
    return """
    <h1>🌍 DisasterShield 360</h1>

    <form action="/predict" method="post">

    Disaster Type:
    <input type="text" name="disaster"><br><br>

    Number of Affected People:
    <input type="number" name="people"><br><br>

    <button type="submit">
    Calculate Resources
    </button>

    </form>
    """


@app.route("/predict", methods=["POST"])
def predict():

    disaster = request.form["disaster"]
    people = int(request.form["people"])

    food = people * 3
    water = people * 5
    medical = max(1, people // 10)
    shelter = max(1, people // 20)

    return f"""
    <h1>🚨 Disaster Resource Report</h1>

    Disaster Type: {disaster}<br><br>

    👥 Affected People: {people}<br><br>

    🍚 Food Packets Required: {food}<br><br>

    💧 Water Bottles Required: {water}<br><br>

    🏥 Medical Kits Required: {medical}<br><br>

    🏠 Shelter Units Required: {shelter}
    """


def run():
    app.run(host="0.0.0.0", port=5000)


Thread(target=run).start()
```
